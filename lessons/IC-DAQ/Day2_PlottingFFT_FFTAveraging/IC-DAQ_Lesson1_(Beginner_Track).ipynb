## Strings
In Python, strings are sequences of characters, and various operations can be performed on them using built-in operators.

```python
# Creating strings
# You can use both double and single quote marks to create a string ("..." and '...')
str1 = 'Hello World!'
str2 = "Hello world"
str3 = '''This is a multiline
string'''
str4 = """This is a multiline
string"""
print(str1)
print(str2)
print(str3)
print(str4)
```

```python
# Accessing individual characters
word = "Python"
print(word[0])  
print(word[-1]) 

# Slicing strings
print(word[1:4])  
print(word[:2])   
print(word[2:])   
print(word[:]) 
```

```python
# Formatting 
resistance = 10.5 
print(f"{resistance} Ohms")

pi = 3.14159
print(f"{pi:.3f}") # 3 decimal places

index = 1
print(f"{index:03d}") # Leading zeros 

num = 123090981374 
print(f"{num:,}") # Comma separator for thousdands 
```

```python
# Combining strings ("concatenation")
resistance1 = f"{10.5}"
resistance2 = str(15.4)
print(resistance1 + ',' + resistance2)
```

# Section 3: Functions

Functions are reusable blocks of code that perform a specific task. They are defined using the 'def' keyword, followed by the function name, parenthesis, and a colon. The code within the function is indented, and the return value(s) are specified with the 'return' keyword.

```python
# define a function
def func():
    print('Xin chào')
```

```python
# execute the function
func()
```

```python
# define a function with a parameter 
def func(name):
    '''
    Greets the person given by the input
    
    Parameters:
        name: str, This name of the person we are greeting.
    Returns:
        None
    '''
    
    print("Xin chào," , name)
```

```python
func("Chris")
```

```python
# define a function with a parameter 
def func(x):
    '''
    Evaluates the quadratic x**2 + 1
    
    Parameters:
        x: float
    Returns:
        float, f(x)=x**2 + 1
    '''
    return x**2 + 1
```

```python
y=func(2)
print(y)
```

```python
# Define a function with two parameters, and return two parameters
def add_subtract_numbers(a, b):
    '''
    Creates a pair of the sum and difference of the input.
    
    Parameters:
        a: float
        b: float
    Returns:
        x: float, sum of a and b
        y: float, difference of a and b
    '''
    x = a + b 
    y = a - b
    return x, y
```

```python
# Call the function 
x, y = add_subtract_numbers(10, 4)
print(f'sum       : {x}')
print(f'difference: {y}')
```

```python
# When you output multiple values, it is actually given as a tuple.
t = add_subtract_numbers(4, 7)
print(f'output    : {t}')
x, y = t
print(f'sum       : {x}')
print(f'difference: {y}')
```

```python
# I can also pass in a list instead of individual parameters using the asterix operator
params = (4, 17)
x, y = add_subtract_numbers(*params)
print(f'sum       : {x}')
print(f'difference: {y}')
```

### Default parameters
Some parameters can be given a default value, that will only be overwritten if a different value is given as input to the function. 

```python
def print_status(instrument, status = 'enabled'):
    print(f'{instrument}: {status}')
print_status('Function generator')
print_status(instrument = 'Function generator')
print_status('Function generator', status = 'disabled')
print_status('Function generator', 'disabled')
```

### Higher order functions
Functions can take other functions as arguments or return them as results.

```python
def square(x):
    return x * x 
def print_func_output(func, x):
    '''
    Prints the return value of func applied to x
    
    parameters:
        func: function, takes x as a parameter
        x: fits data type of func
    returns:
        None
    '''
    print(func(x))
    
print_func_output(square, 2)
```

### Using functions made by someone else: Modules
Python has many built in functions, e.g. print(). However, in order not to include unnecessary things that will slow down your code, not all the functions are immediately available. You need to tell Python to go find the other functions.

**You should learn how to read the documentation for use any function you want to import**

```python
# Option 1: Import an entire module
import numpy
# call functions from that module
print(numpy.sin(3.14159/4))
print(numpy.cos(3.14159/4))
print(0.5*numpy.sqrt(2))
```

```python
# Option 2: Only import the functions that you need
from numpy import sin, cos
print(sin(0))
print(cos(0))
```

It is clear how to use sin and cos, but how about for a function that is not as clear?
Consider this random integer function "randint":

https://numpy.org/doc/stable/reference/random/generated/numpy.random.randint.html

We can see that it takes a lower limit, upper limit, size, and data type. Upper limit, size, and data type all have default parameters. By looking at the documentation, what does this function return?

```python
# Modules can have smaller modules contained in them, called submodules.
from numpy.random import randint # numpy is the module, numpy.random is the submodule

print(randint(0,10))
print(randint(0,10))
print(randint(0,10))
```

# Section 4: Control Flow
Control flow describes the order in which segments of code are executed

## "If-else" statements

```python
x = True
if x:
    print("x is true")
else:
    print("x is false")
```

```python
# if-elif-else statement executes the first condition that is True 
x=2
if x<1:
    print('x is less than 1')
elif x>1: # elif = "else if"
    print('x is greater than 1')
else:
    print('x is 1')
```

## Loops
Loops are blocks of code that continue until a condition is met. This condition is either defined by the user or implied by the loop structure.

```python
# while loop: executes the code as long as a condition is True 
i = 0
while i <= 3:
    print("i =",i)
    i += 1
```

```python
# for loop: loops over each item in a sequence.
exponents = [0, 1, 2, 3]
for i in exponents:
    print(10**i)
```

```python
# for loops can be used to easily generate a list
magnitudes = [10**i for i in exponents]
print(magnitudes)

# if-else statements can be included too
even_magnitudes = [10**i for i in exponents if i%2==0]
# recall that % operator returns the remainder after division
print(even_magnitudes)
```

### Useful functions to use with "for" loops

```python
# range creates evenly spaced integers
r = range(4) # list of integers up to (not including) 4
for index in r:
    print(index)
```

```python
# enumerate: iterate through a list and also get the index of each item 
integers = [10, 6, 5, 9, 3]
for index, integer in enumerate(integers):
    print(index, integer)
```

# Section 5: Review Example
Example: You are writing a function that controls a telescope. You only want to observe when it is night time and the weather is good.

To check the time, use the imported functions from "time" module. To check the weather, suppose you have the function "check_weather()" already written.

```python
from time import ctime

current_date_and_time = ctime()
print(current_date_and_time)

#extract the hour from the string
time_list = current_date_and_time.split(' ')
current_time = time_list[3]
current_hour = int(current_time.split(':')[0])
print(current_hour)
```

```python
from numpy import random

def check_weather():
    random_value = random.randint(0,4) # random seed of either 0, 1, or 2
    if random_value == 0 or random_value == 3:
        return "clear"
    elif random_value == 1:
        return "rain"
    else:
        return "cloudy"
```

```python
# Make a check to see if it is night and the weather is good

# boolean to check the time
is_night = current_hour < 4 or current_hour > 20

# boolean to check weather
is_clear = check_weather() == "clear"

if is_night and is_clear:
    print("Good for observations")
else:
    if not is_night:
        print("Not night")
    else:
        print("Night")
        
    if not is_clear:
        print("Not clear")
    else:
        print("Clear")
```

```python
# Make this check into a function we can reuse
def observe_bool(hour,weather):
    '''
    Given the hour and the weather, determine if we can observe.

    hour: int, current hour
    weather: str, current weather

    return: bool, True if observing conditions are good
    '''
    is_night = hour < 5 or hour > 20

    # boolean to check weather
    is_clear = weather == "clear"
    
    if is_night and is_clear:
        print("Good for observations")
        return True
    else:
        if not is_night:
            print("Not night")
            
        if not is_clear:
            print("Not clear")
        return False
```

```python
# define a dummy function that is the placeholder for taking observations
def use_telescope():
    print("Observing...")

from time import sleep

# Put this in a loop to check every hour
hours = range(7)
for h in hours:
    print("Hour:",h)
    sleep(0.5) # wait 2 seconds
    if observe_bool(h, check_weather()):
        use_telescope()
    print("\n")
```
