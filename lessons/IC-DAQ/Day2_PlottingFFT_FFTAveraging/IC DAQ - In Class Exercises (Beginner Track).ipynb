### Exercise 2

Performing a rotation on a set of points using Numpy. Verify your code by plotting the rotated data.

Recall that a point defined by vector $(x, y)$ can be rotated counterclockwise by a multiplying it with a rotation matrix 

$$
R = \begin{bmatrix}\cos(\theta) & -\sin(\theta) \\ \sin(\theta) & \cos(\theta)\end{bmatrix}
$$

So the rotated point is 
$$
\begin{bmatrix}x' \\ y'\end{bmatrix} =  R \begin{bmatrix}x \\ y\end{bmatrix} = \begin{bmatrix}\cos(\theta) & -\sin(\theta) \\ \sin(\theta) & \cos(\theta)\end{bmatrix} \begin{bmatrix}x \\ y\end{bmatrix}
$$

Let us start by rotating a single point.

```python
def rotate_matrix(angle):
    '''
    Creates a counter clockwise rotation matrix by a given angle
    Parameters:
        angle: float, rotation angle in radians
    Returns:
        2x2 numpy array, rotation matrix

    '''
    return np.array([
                        [np.cos(angle), -np.sin(angle)],
                        [np.sin(angle), np.cos(angle)]
    ])
```

```python
unit_circle_x = np.cos( np.linspace(0, 2*np.pi, 100) ) # create 100 points that lie on the unit circle
unit_circle_y = np.sin( np.linspace(0, 2*np.pi, 100) )


point = np.array([0, 1])

theta = np.pi / 4
R = rotate_matrix(theta)

new_point = np.matmul(R,point) # numpy matrix multiplication

plt.figure()
plt.plot(unit_circle_x,unit_circle_y)
plt.scatter(point[0],point[1],color='red', label="Original Point")
plt.scatter(new_point[0],new_point[1], color='green',label="Rotated Point")
plt.legend()
plt.show()
```

# Linear Regression

Liner regression is an important statistics technique that uses a linear model to describe the relation between 2 or more data sets. The most basic form is **simple linear regression**, which takes 2 data sets $x$ and $y$ and attempts to find the values $a$, $b$ to best fit the model.
$$
y = a*x + b
$$

Here is an example of a data set that simple linear regression would be useful for:

Suppose there was a studied that followed 10 people who smoked cigarettes. It measured how many cigarettes each person smoke per day, and how old they were when they died.


```python
cigarettes_smoked_per_day = [1, 1, 2, 0.5, 4, 2, 1, 3, 5, 1]
age_at_death = [81, 85, 80, 88, 72, 79, 80, 74, 65, 79]

plt.figure()
plt.scatter(cigarettes_smoked_per_day, age_at_death)
plt.xlabel("Cigarettes smoked per day")
plt.ylabel("Age at death (years)")
plt.show()
```

We can see that the data has a strong linear correlation. 

If we found the coefficient $a$ using simple linear regression, it would tell us how many years you lose per daily cigarette.
The value $b$ would tell us the average age at death for people who do not smoke.

Linear regression works by finding a line of best fit to the data. A "line of best fit" means that if you calcaluated how far each real data point was from the theoretical line, the total distance of all the points to the line is minimized. 

### Mathematical Framework

Let $x_i=\{x_i\}_{i=1}^{n}$ be the observed data for the independent variable x. Let $y_i=\{y_i\}_{i=1}^{n}$ be the observed data for the variable we want to predict (dependent variable).

We have already seen our theoretical model
$$
y = a*x + b
$$

This model predicts that if we have the data set $x_i$, we should observe
$$
y = a*{x_i }+ b
$$

We can then define the error $\hat{e_i}$ between the predicted and observed values:
$$
\hat{e_i} = y_i - y
$$
$$
\hat{e_i} = y_i - a*{x_i } - b
$$

We want to minimize this error term to make the model be as close to the data as possible. Since we only want the total distance (we do not care if it is positive or negative), we use the square of the error time to define an objective function $Q$ that calculates this squared distance:
$$
Q = \sum_{i=1}^n \hat{e_i}^2
$$
$$
Q = \sum_{i=1}^n (y_i - a*{x_i } - b)^2
$$

Thus, we want to find values $\hat{a}$, $\hat{b}$ that minimize $Q$.

Now we know the outline of what simple linear regression is trying to do. We will skip the details of the derivation, but after solving for the minimzing inputs of $Q$, we get these results. Recall that $n$ is the length of $x_i$.

$$
\hat{a} = \frac{n \sum (x_i * y_i)  - \sum x_i \sum y_i}{ n \sum x_i^2 - (\sum x_i)^2}
$$

$$
\hat{b} = \frac{\sum y_i \sum x_i^2 - \sum x_i \sum (x_i * y_i)}{ n \sum x_i^2 - (\sum x_i)^2}
$$
