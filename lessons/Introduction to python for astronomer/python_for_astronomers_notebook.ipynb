## 8  Plottings

```python
# The all-powerful matplotlib library
# First example: 1D bell curve
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------
# Parameters for the 1-D Gaussian
# -------------------------------------------------
mu     = 0.0     # mean (centre)
sigma  = 1.0     # standard deviation (width)

# Generate x values (cover ±4σ is plenty)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 400)

# Calculate corresponding y values for the Gaussian
y = np.exp(-((x - mu)**2) / (2 * sigma**2))

# -------------------------------------------------
# Create the plot
# -------------------------------------------------
plt.figure(figsize=(8, 6))
plt.plot(x, y, 'b*', label='1D Gaussian')   # blue line

# Labels and title
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.title('1D Gaussian ')

# Grid and legend
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# The all-powerful matplotlib library
# Second example: 2D bell curve
import numpy as np
import matplotlib.pyplot as plt
def gaussian_2d(x, y, mu_x, mu_y, sigma_x, sigma_y, rho):
    """
    Calculates the value of a 2D Gaussian probability density function.

    Parameters:
    x, y (arrays): Coordinates at which to evaluate the Gaussian.
    mu_x, mu_y (float): Means of the x and y dimensions.
    sigma_x, sigma_y (float): Standard deviations of the x and y dimensions.
    rho (float): Correlation coefficient between x and y.
    """
    z = (x - mu_x)**2 / sigma_x**2 - 2 * rho * (x - mu_x) * (y - mu_y) / (sigma_x * sigma_y) + (y - mu_y)**2 / sigma_y**2
    return (1 / (2 * np.pi * sigma_x * sigma_y * np.sqrt(1 - rho**2))) * np.exp(-z / (2 * (1 - rho**2)))

# Define parameters for the 2D Gaussian
mu_x, mu_y = 0, 0  # Mean
sigma_x, sigma_y = 1, 2  # Standard deviations
rho = 0.5  # Correlation coefficient

# Create a grid of points
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Calculate the 2D Gaussian values
Z = gaussian_2d(X, Y, mu_x, mu_y, sigma_x, sigma_y, rho)


# Plotting as a 3D surface
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_title('3D Surface Plot of 2D Gaussian')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Probability Density')
plt.show()
```

```python
# The all-powerful matplotlib library
# Third example: Grid of plots
import matplotlib.pyplot as plt
import numpy as np

# Create some sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = x**2
y4 = np.exp(-x)

# Create a 2x2 grid of subplots
# fig is the Figure object, axs is a 2D array of Axes objects
fig, axs = plt.subplots(2, 2, figsize=(8, 6)) # figsize sets the figure size in inches

# Plot in the top-left subplot (row 0, column 0)
axs[0, 0].plot(x, y1, color='blue')
axs[0, 0].set_title('Sine Wave')
axs[0, 0].set_xlabel('X-axis')
axs[0, 0].set_ylabel('Y-axis')

# Plot in the top-right subplot (row 0, column 1)
axs[0, 1].plot(x, y2, color='green')
axs[0, 1].set_title('Cosine Wave')
axs[0, 1].set_xlabel('X-axis')
axs[0, 1].set_ylabel('Y-axis')

# Plot in the bottom-left subplot (row 1, column 0)
axs[1, 0].plot(x, y3, color='red')
axs[1, 0].set_title('Quadratic Function')
axs[1, 0].set_xlabel('X-axis')
axs[1, 0].set_ylabel('Y-axis')

# Plot in the bottom-right subplot (row 1, column 1)
axs[1, 1].plot(x, y4, color='purple')
axs[1, 1].set_title('Exponential Decay')
axs[1, 1].set_xlabel('X-axis')
axs[1, 1].set_ylabel('Y-axis')

# Adjust layout to prevent overlapping titles/labels
plt.tight_layout()

# Display the figure
plt.show()
```

## 9  File I/O saving and reading

```python
# numpy savetxt allows us to save and read file
# Recall the example of 1D bell curve. We save x and y coordinates as a text file with header
import numpy as np

# -------------------------------------------------
# Parameters for the 1-D Gaussian
# -------------------------------------------------
mu     = 0.0     # mean (centre)
sigma  = 1.0     # standard deviation (width)

# Generate x values (cover ±4σ is plenty)
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 400)

# Calculate corresponding y values for the Gaussian
y = np.exp(-((x - mu)**2) / (2 * sigma**2))

# Stack the columns and save
data = np.column_stack((x, y))
np.savetxt(
    'gaussian_xy.txt',   # output file name
    data,                # 2-column array
    header='x y',        # column headers
    fmt='%.8f',          # numeric format (8 decimal places)
    comments=''          # prevent '#' in front of header
)

print("Saved to gaussian_xy.txt")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.animation import FuncAnimation
import imageio
import os

frame_start = 148
frame_end = 165

# Path to the video file on your COMPUTER or Colab
video_path = '/content/sample_data/EBC_local_250124_143721.MP4'

# Open the video file
cap = cv2.VideoCapture(video_path)

# Check if video was successfully opened
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

# Get video properties
fps = cap.get(cv2.CAP_PROP_FPS)  # Frames per second
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))  # Total number of frames

print(f"Video FPS: {fps}")
print(f"Total Frames: {frame_count}")

# Initialize a list to store frames
frames = []

# Loop through all frames in range
for frame_num in range(frame_start, frame_end):
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_num)
    ret, frame = cap.read()

    if not ret:
        print(f"Error: Could not read frame {frame_num}.")
        continue

    # Convert the frame to grayscale (single channel)
    #frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    frame_gray = frame[:,:,0]
    # Append the grayscale frame to the list
    frames.append(frame_gray)


# Convert the list of frames to a numpy array
frames_np = np.array(frames)  # Shape will be (frame_count, height, width)
print(f"Shape of the numpy array: {frames_np.shape}")
```
