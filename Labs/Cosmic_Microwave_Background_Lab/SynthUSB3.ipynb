```python
import numpy as np
import serial
import time
```

Find the COM port that is assigned to the USB Synth and use it in the cell below.

```python
# Set up serial connection to COM5
ser = serial.Serial('COM5', baudrate=9600, timeout=1)

# Make sure the connection is open
if ser.is_open:
    print("Serial connection established.")
```

```python
# Set frequency in MHz using b'f...'
# Set power in dBm using b'W...'
# The "b" converts the string into a byte string that the serial connection can transmit.
ser.write(b'f1200')
ser.write(b'W-20.0')

# Query the current frequncy 
ser.write(b'f?')

# Read the response
response = ser.readline().decode('utf-8').strip()
print("Current Freq:", response)

# Query the current frequncy 
ser.write(b'W?')

# Read the response
response = ser.readline().decode('utf-8').strip()
print("Current Power:", response)

# Close the serial connection
ser.close()
```
